```python
# generate report
class DataAnalysis:

    def generate_report(self):
        self.df = pd.read_csv("library_transaction.csv")

        print("----- Library Transaction Report -----")
        print("Total Transactions:", len(self.df))
        print("Total Users:", self.df["User_ID"].nunique())
        print("Total Books:", self.df["Book_Title"].nunique())
        print("Average Borrowing Duration:", self.df["Borrowing_Duration_Days"].mean())
```

## 3.Data Analysis and computations

```python
class DataAnalysis:

    def calculate_duration_statistics(self):
        df = pd.read_csv("library_transaction.csv")

        duration = df["Borrowing_Duration_Days"]

        print("Borrowing Duration Statistics")
        print("Mean:", np.mean(duration))
        print("Median:", np.median(duration))
        print("Minimum:", np.min(duration))
        print("Maximum:", np.max(duration))
        print("Standard Deviation:", np.std(duration))
```

```python
obj=DataAnalysis()
obj.calculate_duration_statistics()
```

## 4.Data Visualization

```python
# bar chart
df = pd.read_csv("library_transaction.csv")

top_5_books = df["Book_Title"].value_counts().head(5)

plt.bar(top_5_books.index, top_5_books.values)

plt.title("Top 5 Most Borrowed Books")
plt.xlabel("Book Title")
plt.ylabel("Number of Borrowings")
plt.xticks(rotation=45)
plt.show()
```

```python
# line graph


df = pd.read_csv("library_transaction.csv")

df["Date"] = pd.to_datetime(df["Date"])

monthly_borrowing = df.groupby(df["Date"].dt.to_period("M")).size()

plt.plot(monthly_borrowing.index.astype(str), monthly_borrowing.values, marker="o")

plt.title("Borrowing Trend Over Months")
plt.xlabel("Month")
plt.ylabel("Number of Borrowings")
plt.xticks(rotation=45)
plt.show()
```

```python
# pie chart:

df = pd.read_csv("library_transaction.csv")

genre_count = df["Genre"].value_counts()

plt.pie(
    genre_count.values,
    labels=genre_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Books Borrowed by Genre")
plt.show()
```

```python
# heatmap

df = pd.read_csv("library_transaction.csv")

df["Date"] = pd.to_datetime(df["Date"])

df["Day"] = df["Date"].dt.day_name()

heatmap_data = pd.crosstab(
    df["Day"],
    df["Date"].dt.month
)

plt.figure(figsize=(10, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Borrowing Activity by Day and Month")
plt.xlabel("Month")
plt.ylabel("Day")

plt.show()
```
